## **QueryMart - Retail Sales Database**


```python
# Step 1: Import Required Libraries
import sqlite3
import requests
import pandas as pd
```

```python
import requests
url = "https://github.com/YBIFoundation/SQL/raw/main/QueryMart.db"
response = requests.get(url)
with open("QueryMart.db", "wb") as f:
    f.write(response.content)
conn = sqlite3.connect("QueryMart.db")
print("Database downloaded successfully")
```

```python
# Create helper function
def sql(query):
    return pd.read_sql(query, conn)
```

```python
sql(" SELECT name FROM sqlite_master WHERE type='table';")
```

```python
# Customer Table
sql("SELECT * FROM customers;")
```

```python
# Product Table
sql("SELECT * FROM products")
```

```python
# Orders Table
sql("SELECT * FROM order_items")
```

```python
# Calculate Total Revenue
sql("SELECT SUM(net_amount) FROM order_items")
```

```python
sql("SELECT * FROM orders")
```

```python
# Find unique payment methods
sql("SELECT DISTINCT payment_method	FROM orders")
```

```python
# Find unique order statuses
sql("SELECT DISTINCT order_status FROM orders")
```

```python
#Find all delivered orders
sql("SELECT * FROM orders WHERE order_status = 'Delivered'")
```

```python
# Find all orders placed after June 2025
sql("SELECT * FROM orders WHERE order_date >= '2025-07-01'")
```

```python
# Find orders placed between January and March 2025
sql("SELECT * FROM orders WHERE order_date BETWEEN 	'2025-01-01' AND 	'2025-03-31'")
```

```python
# Find the 10 most expensive products
sql("SELECT * FROM products ORDER BY unit_price DESC LIMIT 10")
```

```python
# Calculate average order-item revenue
sql("SELECT AVG(net_amount) FROM order_items")
```

```python
# Find number of products in each category
sql("SELECT category, COUNT(*), AVG(unit_price) FROM products GROUP BY category")
```

```python
# 1. Product Catalog - Find the total number of products.
sql("SELECT COUNT(*) AS total_products FROM products")
```

```python
# 2. Category Analysis - Find the number of products in each category.
sql("SELECT category, COUNT(*) AS num_products FROM products GROUP BY category")
```

```python
# 3. Price Analysis - Find the 5 most expensive products.
sql("SELECT product_id, product_name, unit_price FROM products ORDER BY unit_price DESC LIMIT 5")
```

```python
# 4. Premium Products - Find all products with a price greater than ₹10,000.
sql("SELECT product_id, product_name, category, unit_price FROM products WHERE unit_price > 10000")
```
